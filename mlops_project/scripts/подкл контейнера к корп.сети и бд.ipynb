```python
import socket
import os
from dotenv import load_dotenv

load_dotenv('.env')

# Тестируем доступность сервера
server = "172.16.0.108"  # или "db23.vra.local"
port = 1433

print(f"Проверка доступа к {server}:{port}...")

sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
sock.settimeout(5)
result = sock.connect_ex((server, port))
if result == 0:
    print(f"✅ Порт {port} открыт! Контейнер видит сервер.")
else:
    print(f"❌ Порт {port} НЕ ДОСТУПЕН (код ошибки: {result})")
sock.close()

# Теперь пробуем подключиться к БД
import pymssql

user = os.getenv('DB_USER').replace('\\\\', '\\')
password = os.getenv('DB_PASSWORD')
database = os.getenv('DB_NAME')

try:
    conn = pymssql.connect(
        server=server,  # или "172.16.0.108"
        user=user,
        password=password,
        database=database,
        timeout=10
    )
    print("✅ Подключение к БД успешно!")
    conn.close()
except Exception as e:
    print(f"❌ Ошибка: {e}")
```
